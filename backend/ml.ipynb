```python
from huggingface_hub import hf_hub_download
from llama_cpp import Llama
```

```python
import os
```

```python
model_name = "unsloth/gemma-3-4b-it-GGUF"
model_file = "gemma-3-4b-it-Q4_K_M.gguf"

if not os.path.exists(model_file):
    print("Downloading model...")
    hf_hub_download(
        repo_id=model_name,
        filename=model_file,
        local_dir=".",
        resume_download=True
    )
else:
    print("Model already exists")


print("\nModel status:", "Found" if os.path.exists(model_file) else "Missing")
```

```python
try:

    llm = Llama(
        model_path=model_file,
        n_ctx=16384,
        n_gpu_layers=-1,
        verbose=False
    )
    print("LLM initialized successfully!")

except Exception as e:
    print(f"LLM initialization failed: {str(e)}")
```

```python
def generate_data(API,age,disease):
    try:
        prompt = f"""[INST] You are a AQI prevention solution provider. the current AQI is {API}. give 5 precaution to take according the age of the person is{age} with the disease which is{disease} [/INST]"""

        output = llm(
            prompt,
            max_tokens=2000,
            temperature=0.8,
            echo=False
        )
        answer = output['choices'][0]['text'].strip()
        print(answer)
        return answer

    except Exception as e:
        print(f"Error: {str(e)}")
        return "Could not generate Chapters"
```

```python
generate_data(128,18,"fever")
```
